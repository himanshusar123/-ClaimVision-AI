```python
def employee_status(age):
    if age >= 18:
        return "Eligible"

    return "Not Eligible"
```

```python
def divide(a, b):
    return a / b
```

```python
import pytest

def test_division():

    assert divide(10, 2) == 5

def test_division_by_zero():

    with pytest.raises(ZeroDivisionError):
        divide(10, 0)
```

```python
import pytest

@pytest.fixture
def employee_data():

    return {
        "employee_id": 101,
        "name": "Rahul",
        "department": "IT",
        "salary": 50000
    }
```

```python
import pytest

@pytest.mark.parametrize(
    "salary, expected",
    [
        (10000, 1000),
        (20000, 2000),
        (30000, 3000),
        (50000, 5000),
        (100000, 10000)
    ]
)

def test_calculate_bonus(salary, expected):

    assert calculate_bonus(salary) == expected
```

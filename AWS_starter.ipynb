```python
import boto3
import json
from botocore.exceptions import BotoCoreError, ClientError

def generate_identity_report():
    try:
        sts = boto3.client("sts")
        identity = sts.get_caller_identity()

        print("AWS Account:", identity["Account"])

        with open("identity_audit.json", "w") as file:
            json.dump(identity, file, indent=4)

        return True

    except (BotoCoreError, ClientError) as error:
        print("AWS request failed:", error)
        return False

print("Success:", generate_identity_report())
```

```python
import boto3
import json
from botocore.exceptions import BotoCoreError, ClientError

def aws_environment_check():
    try:
        sts = boto3.client("sts")
        identity = sts.get_caller_identity()

        report = {
            "status": "SUCCESS",
            "account": identity["Account"],
            "arn": identity["Arn"]
        }

        with open("aws_environment_check.json", "w") as file:
            json.dump(report, file, indent=4)

        print("AWS environment check: SUCCESS")
        return report

    except (BotoCoreError, ClientError) as error:
        print("AWS environment check: FAILED")
        return {"status": "FAILED", "error": str(error)}

result = aws_environment_check()
print(result)
```

```python
import boto3
import json
from botocore.exceptions import BotoCoreError, ClientError

TARGET_ACCOUNT = "REPLACE_WITH_APPROVED_ACCOUNT"

def account_guard():
    try:
        sts = boto3.client("sts")
        identity = sts.get_caller_identity()
        actual_account = identity["Account"]

        if actual_account == TARGET_ACCOUNT:
            status = "APPROVED"
            message = "Target AWS account verified."
        else:
            status = "BLOCKED"
            message = "AWS account does not match target account."

        result = {
            "status": status,
            "account": actual_account,
            "arn": identity["Arn"],
            "message": message
        }

        with open("account_guard.json", "w") as file:
            json.dump(result, file, indent=4)

        print(message)
        return result

    except (BotoCoreError, ClientError) as error:
        print("AWS verification failed:", error)
        return {"status": "ERROR", "error": str(error)}

result = account_guard()
print(result)
```

```python
import boto3
from botocore.exceptions import BotoCoreError, ClientError

def aws_identity_summary():
    try:
        # Create STS client
        sts = boto3.client("sts")

        # Get current AWS identity
        identity = sts.get_caller_identity()

        # Build and return a clean result
        return {
            "account": identity["Account"],
            "arn": identity["Arn"],
            "status": "VERIFIED"
        }

    except (BotoCoreError, ClientError) as error:
        # Return a structured error result
        return {
            "account": None,
            "arn": None,
            "status": "ERROR",
            "message": str(error)
        }

print(aws_identity_summary())
```

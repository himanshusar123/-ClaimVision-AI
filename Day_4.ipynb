```python
# # File Handling – TXT, CSV & JSON

# Suppose I give you a CSV file containing 10,000 customer records.
# Can we manually put all those records into Python variables?"
```

```python
# TXT File Handling
# Reading a File
file = open("sample.txt", "r")

content = file.read()

print(content)

file.close()
```

```python
# Better Approach-With

with open("sample.txt", "r") as file:
    content = file.read()

print(content)
```

```python
# Writing to a File

with open("output.txt", "w") as file:
    file.write("AWS Python Training")
```

```python
with open("output.txt", "r") as file:
    print(file.read())
```

```python
# CSV Files

import csv
with open("employees.csv", "r") as file:

    reader = csv.DictReader(file)

    for row in reader:
        print(row)
```

```python
# CSV Filtering Example
import csv

with open("employees.csv", "r") as file:

    reader = csv.DictReader(file)

    for row in reader:

        if row["department"] == "IT":
            print(row["name"])
```

```python
# JSON Files
{
    "name": "server-01",
    "region": "ap-south-1",
    "status": "running"
}
```

```python
import json

with open("server.json", "r") as file:
    data = json.load(file)

print(data)
```

```python
print(data["name"])
print(data["region"])
print(data["status"])
```

```python
# Nested JSON
{
    "server": {
        "name": "server-01",
        "region": "ap-south-1",
        "configuration": {
            "cpu": 4,
            "memory": 16
        }
    }
}
```

```python
import json

with open("server.json", "r") as file:
    data = json.load(file)

print(data["server"]["name"])
print(data["server"]["region"])
print(data["server"]["configuration"]["cpu"])
```

```python
# JSON Writing
import json

server = {
    "name": "server-01",
    "region": "ap-south-1",
    "status": "running"
}

with open("server_output.json", "w") as file:
    json.dump(server, file, indent=4)
```

```python
# Exception Handling with files

try:

    with open("missing.txt", "r") as file:
        data = file.read()

except FileNotFoundError:

    print("File does not exist")
```

```python
# Multiple Possible Errors

try:

    with open("data.json", "r") as file:
        data = json.load(file)

except FileNotFoundError:

    print("File not found")

except json.JSONDecodeError:

    print("Invalid JSON format")
```

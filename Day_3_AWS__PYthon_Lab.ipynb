```python
# List Comprehensions
files=["sales.csv","customer.csv","orders.csv"]
upper_files=[]
for files in files:
    upper_files.append(files.upper())
print(upper_files)
```

```python
files=["sales.csv","customer.csv","orders.csv"]
upper_files=[files.upper() for files in files]
print(upper_files)
#
```

```python
# Conditional list comprehension

files=["sales.csv","customer.txt","orders.csv"]
upper_files=[files.upper() for files in files if files.endswith(".csv")]
print(upper_files)
#
```

```python
# Exception Handling

# What happens when our Python program encounters something unexpected?"

number=10
divisor=0
result=number/divisor
print(result)
```

```python
try:
    result=number/divisor
    print(result)
except ZeroDivisionError:
    print("Cannot divide by Zero")
```

```python
# try

# Contains the code that might produce an error.

# except

# Contains what the program should do when that particular error occurs.

try:
  number=int("abc")
except ValueError:
  print("Invalid number")
```

```python
#finally

# Finally is used when we want something to execute regardless of whether an error occurred.


try:
    number = int("100")
    print(number)

except ValueError:
    print("Invalid input")

finally:
    print("Processing completed")
```

```python
resources = [
    {"name": "server-01", "status": "running"},
    {"name": "server-02", "status": "stopped"},
    {"name": "server-03", "status": "running"}
]

for resource in resources:

    try:
        if resource["status"] == "running":
            print(resource["name"], "is running")

    except KeyError:
        print("Status information is missing")
```

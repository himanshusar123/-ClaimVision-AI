```python
# A cloud data-processing team receives multiple files every day.
#  We want to check the size of each file and classify it as small,
#   medium or large

def check_file(file_name,file_size):
  if file_size>100:
    print(file_name,"is a large file")
  elif file_size>50:
    print(file_name,"is a medium file")
  else:
    print(file_name,"is a small file")
```

```python
check_file("sales.csv",75)
```

```python
check_file("customer.csv",120)
```

```python
check_file("products.csv",30)
```

```python
# Combine Function with a Loop
files = [
    ("sales.csv", 75),
    ("customers.csv", 120),
    ("products.csv", 30)
]

for file_name, file_size in files:
    check_file(file_name, file_size)
```

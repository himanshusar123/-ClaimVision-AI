```python
import json

def lambda_handler(event, context=None):
    for record in event.get("Records", []):
        bucket = record["s3"]["bucket"]["name"]
        key = record["s3"]["object"]["key"]
        event_name = record.get("eventName", "Unknown")

        print("S3 event received")
        print("Bucket:", bucket)
        print("Object:", key)
        print("Event:", event_name)

    return {
        "statusCode": 200,
        "body": json.dumps({
            "message": "S3 event processed"
        })
    }
```

```python
import json

def lambda_handler(event, context=None):
    file_name = event.get("file_name")
    file_type = event.get("file_type", "unknown")

    if not file_name:
        result = {
            "status": "ERROR",
            "message": "file_name is required"
        }
        return {
            "statusCode": 400,
            "body": json.dumps(result)
        }

    result = {
        "status": "ACCEPTED",
        "file_name": file_name,
        "file_type": file_type
    }

    return {
        "statusCode": 200,
        "body": json.dumps(result)
    }

# Test with valid input
print(lambda_handler({
    "file_name": "report.csv",
    "file_type": "csv"
}))

# Test with missing input
print(lambda_handler({}))
```

```python
import json
from urllib.parse import unquote_plus

def lambda_handler(event, context=None):
    records = event.get("Records", [])
    processed = []

    for record in records:
        s3_info = record.get("s3", {})
        bucket = s3_info.get("bucket", {}).get("name")
        key = unquote_plus(
            s3_info.get("object", {}).get("key", "")
        )

        processed.append({
            "bucket": bucket,
            "key": key,
            "event_name": record.get("eventName", "Unknown")
        })

    print("S3 upload event:", json.dumps(processed, indent=2))

    return {
        "statusCode": 200,
        "body": json.dumps({
            "processed_records": processed
        })
    }

sample_event = {
    "Records": [
        {
            "eventName": "ObjectCreated:Put",
            "s3": {
                "bucket": {
                    "name": "amzn-demo-bucket21236587"
                },
                "object": {
                    "key": "incoming/sample.txt"
                }
            }
        }
    ]
}

response = lambda_handler(sample_event)
print("\nResponse:")
print(json.dumps(response, indent=2))
```

```python
import json
from urllib.parse import unquote_plus

def lambda_handler(event, context=None):
    records = event.get("Records", [])

    if not records:
        return {
            "statusCode": 400,
            "body": json.dumps({
                "message": "No S3 records found"
            })
        }

    processed = []

    for record in records:
        try:
            s3_info = record["s3"]
            bucket = s3_info["bucket"]["name"]
            key = unquote_plus(s3_info["object"]["key"])

            processed.append({
                "bucket": bucket,
                "key": key
            })

        except (KeyError, TypeError) as error:
            print("Malformed record:", str(error))

    return {
        "statusCode": 200,
        "body": json.dumps({"processed": processed})
    }

print("Test 1 — Empty event")
print(lambda_handler({}))

print("\nTest 2 — Valid S3 event")
print(lambda_handler({
    "Records": [
        {
            "s3": {
                "bucket": {"name": "amzn-demo-bucket21236587"},
                "object": {"key": "incoming/sample.txt"}
            }
        }
    ]
}))
```

```python
import json
from urllib.parse import unquote_plus

def upload_audit_logger(event, context=None):
    records = event.get("Records", [])
    audit_records = []

    for record in records:
        try:
            s3_info = record["s3"]
            bucket = s3_info["bucket"]["name"]
            obj = s3_info["object"]

            audit = {
                "bucket": bucket,
                "key": unquote_plus(obj["key"]),
                "event": record.get("eventName", "Unknown"),
                "size": obj.get("size", "Not provided"),
                "status": "PROCESSED"
            }

            audit_records.append(audit)
            print("AUDIT:", json.dumps(audit))

        except (KeyError, TypeError) as error:
            print("Unable to process record:", str(error))

    return {
        "statusCode": 200,
        "body": json.dumps({
            "records_processed": len(audit_records),
            "audit_records": audit_records
        })
    }

# Simulate two uploaded files
test_event = {
    "Records": [
        {
            "eventName": "ObjectCreated:Put",
            "s3": {
                "bucket": {"name": "amzn-demo-bucket21236587"},
                "object": {
                    "key": "incoming/sales.csv",
                    "size": 250
                }
            }
        },
        {
            "eventName": "ObjectCreated:Put",
            "s3": {
                "bucket": {"name": "amzn-demo-bucket21236587"},
                "object": {
                    "key": "incoming/customers.csv",
                    "size": 500
                }
            }
        }
    ]
}

result = upload_audit_logger(test_event)
print("\nFinal report:")
print(json.dumps(json.loads(result["body"]), indent=2))
```

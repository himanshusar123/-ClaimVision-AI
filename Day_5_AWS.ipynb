```python
import boto3

s3 = boto3.client("s3", region_name="us-east-1")

bucket_name = "amzn-demo-bucket21236587"

response = s3.list_objects_v2(
    Bucket=bucket_name,
    MaxKeys=10
)

print("S3 listing request successful.")

for obj in response.get("Contents", []):
    print(obj["Key"])
```

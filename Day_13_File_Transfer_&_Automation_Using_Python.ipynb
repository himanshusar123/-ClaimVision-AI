```python
import csv
import json
from pathlib import Path

folder = Path("module7_file_automation")
folder.mkdir(exist_ok=True)

# Create a CSV file
sales = [
    {
        "order_id": "ORD101",
        "product": "Keyboard",
        "quantity": 2,
        "unit_price": 1200
    },
    {
        "order_id": "ORD102",
        "product": "Mouse",
        "quantity": 3,
        "unit_price": 500
    },
    {
        "order_id": "ORD103",
        "product": "Monitor",
        "quantity": 1,
        "unit_price": 8500
    }
]

with open(folder / "daily_sales.csv",
          "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(
        file,
        fieldnames=[
            "order_id", "product", "quantity", "unit_price"
        ]
    )
    writer.writeheader()
    writer.writerows(sales)

# Create a JSON file
metadata = {
    "source": "daily_sales",
    "record_count": len(sales),
    "status": "created"
}

(folder / "metadata.json").write_text(
    json.dumps(metadata, indent=2),
    encoding="utf-8"
)

# Create a text file
(folder / "readme.txt").write_text(
    "Module 7: Python file transfer and automation.",
    encoding="utf-8"
)

print("Files created successfully.")
```

```python
import csv
from pathlib import Path

def validate_sales_file(file_path):
    path = Path(file_path)

    if not path.exists():
        return False, "File does not exist."

    if path.stat().st_size == 0:
        return False, "File is empty."

    required = {
        "order_id",
        "product",
        "quantity",
        "unit_price"
    }

    try:
        with open(path, newline="", encoding="utf-8") as file:
            reader = csv.DictReader(file)

            if not reader.fieldnames:
                return False, "CSV headers are missing."

            missing = required - set(reader.fieldnames)

            if missing:
                return False, f"Missing columns: {missing}"

            rows = list(reader)

        if not rows:
            return False, "CSV has no data rows."

        seen_ids = set()

        for row_number, row in enumerate(rows, start=2):
            order_id = row["order_id"].strip()
            product = row["product"].strip()

            if not order_id or not product:
                return False, f"Blank required value at row {row_number}."

            if order_id in seen_ids:
                return False, f"Duplicate order ID at row {row_number}."

            seen_ids.add(order_id)

            try:
                quantity = int(row["quantity"])
                unit_price = float(row["unit_price"])
            except (ValueError, TypeError):
                return False, f"Invalid numeric value at row {row_number}."

            if quantity <= 0 or unit_price < 0:
                return False, f"Invalid quantity or price at row {row_number}."

        return True, f"Validation passed: {len(rows)} data rows."

    except (OSError, csv.Error, UnicodeError) as error:
        return False, f"Unable to read CSV: {error}"


valid, message = validate_sales_file(
    "module7_file_automation/daily_sales.csv"
)

print(message)
```

```python
%%writefile validate_sales.py

import csv
import os

def validate_sales_file(file_path):
    """
    Validate a sales CSV file.
    Checks file existence, required columns,
    empty values, and numeric sales amounts.
    """

    required_columns = [
        "order_id",
        "product",
        "quantity",
        "unit_price"
    ]

    if not os.path.exists(file_path):
        return False, f"File not found: {file_path}"

    try:
        with open(file_path, "r", newline="", encoding="utf-8-sig") as file:
            reader = csv.DictReader(file)

            if not reader.fieldnames:
                return False, "CSV file is empty or has no headers."

            missing_columns = [
                column for column in required_columns
                if column not in reader.fieldnames
            ]

            if missing_columns:
                return False, (
                    f"Missing required columns: {missing_columns}"
                )

            row_count = 0

            for row_number, row in enumerate(reader, start=2):
                row_count += 1

                if any(
                    not row.get(column) or not row[column].strip()
                    for column in required_columns
                ):
                    return False, (
                        f"Empty required value at row {row_number}"
                    )

                try:
                    quantity = float(row["quantity"])
                    unit_price = float(row["unit_price"])

                    if quantity <= 0 or unit_price < 0:
                        return False, (
                            f"Invalid quantity or unit price "
                            f"at row {row_number}"
                        )

                except ValueError:
                    return False, (
                        f"Non-numeric quantity or unit price "
                        f"at row {row_number}"
                    )

            if row_count == 0:
                return False, "CSV contains headers but no data rows."

        return True, f"Validation successful. Rows checked: {row_count}"

    except (OSError, csv.Error) as error:
        return False, f"File validation failed: {error}"
```

```python
import pandas as pd
import logging
import os

# Configure logging
logging.basicConfig(
    filename="/content/day13_automation.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    force=True
)

def validate_sales_file(file_path):
    try:
        if not os.path.exists(file_path):
            raise FileNotFoundError(f"File not found: {file_path}")

        df = pd.read_csv(file_path)

        required_columns = ["OrderID", "Product", "Quantity", "Price"]

        missing_columns = [
            col for col in required_columns
            if col not in df.columns
        ]

        if missing_columns:
            logging.error(f"Missing columns: {missing_columns}")
            return False, f"Missing columns: {missing_columns}"

        if df.empty:
            logging.warning("CSV file contains no data rows.")
            return False, "CSV file contains no data rows."

        if df[required_columns].isnull().any().any():
            logging.error("Missing values found in required columns.")
            return False, "Missing values found in required columns."

        logging.info(
            f"Validation successful. Rows checked: {len(df)}"
        )
        return True, f"Validation successful. Rows checked: {len(df)}"

    except FileNotFoundError as e:
        logging.error(str(e))
        return False, str(e)

    except pd.errors.ParserError as e:
        logging.error(f"CSV parsing error: {e}")
        return False, "Invalid CSV format."

    except Exception as e:
        logging.exception("Unexpected validation error")
        return False, f"Unexpected error: {e}"
```

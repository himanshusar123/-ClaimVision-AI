```python
class ApplicationError(Exception):
    pass


class TicketNotFoundError(ApplicationError):
    pass


def get_ticket(ticket_id):

    tickets = {
        101: "Open",
        102: "Closed"
    }

    if ticket_id not in tickets:
        raise TicketNotFoundError(
            "Ticket not found"
        )

    return tickets[ticket_id]


try:

    print(get_ticket(105))

except TicketNotFoundError as e:

    print("Error:", e)
```

```python
%%writefile billing.py

def calculate_bill(
    price: float,
    quantity: int,
    discount: float = 0
) -> float:

    if not isinstance(price, (int, float)) or isinstance(price, bool):
        raise TypeError("Price must be numeric")

    if not isinstance(quantity, int) or isinstance(quantity, bool):
        raise TypeError("Quantity must be an integer")

    if not isinstance(discount, (int, float)) or isinstance(discount, bool):
        raise TypeError("Discount must be numeric")

    if price <= 0:
        raise ValueError("Price must be greater than zero")

    if quantity <= 0:
        raise ValueError("Quantity must be greater than zero")

    if not 0 <= discount <= 100:
        raise ValueError("Discount must be between 0 and 100")

    total = price * quantity

    discount_amount = total * discount / 100

    return total - discount_amount
```

```python
import time

def fetch_order(customer):
    print(f"Fetching order for {customer}...")
    time.sleep(2)
    print(f"Order fetched for {customer}")

start = time.time()

fetch_order("Customer A")
fetch_order("Customer B")
fetch_order("Customer C")

print("Total time:", round(time.time() - start, 2), "seconds")
```

```python
import asyncio
import time

async def fetch_order(customer):
    print(f"Fetching order for {customer}...")
    await asyncio.sleep(2)
    print(f"Order fetched for {customer}")

async def main():
    start = time.time()

    await asyncio.gather(
        fetch_order("Customer A"),
        fetch_order("Customer B"),
        fetch_order("Customer C")
    )

    print("Total time:", round(time.time() - start, 2), "seconds")

await main()
```

```python
def get_delivery_status(order_id):

    print("Calling external delivery API...")

    # Simulating an API response
    return "Delivered"


def customer_response(order_id):

    status = get_delivery_status(order_id)

    if status == "Delivered":
        return "Your order has been delivered."

    elif status == "In Transit":
        return "Your order is on its way."

    else:
        return "Unable to find your order."
```

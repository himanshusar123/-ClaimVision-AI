```python
import boto3

region = "us-east-1"
queue_name = "day12-sqs-ticket-queue"

sqs = boto3.client("sqs", region_name=region)

response = sqs.get_queue_url(QueueName=queue_name)
queue_url = response["QueueUrl"]

print("Queue URL:", queue_url)
```

```python
import json

ticket = {
    "ticket_id": "T-1004",
    "issue": "Unable to connect to VPN",
    "priority": "High",
    "department": "IT Support"
}

send_response = sqs.send_message(
    QueueUrl=queue_url,
    MessageBody=json.dumps(ticket)
)

print("Ticket sent successfully!")
print("Message ID:", send_response["MessageId"])
```

```python
receive_response = sqs.receive_message(
    QueueUrl=queue_url,
    MaxNumberOfMessages=1,
    WaitTimeSeconds=10
)

messages = receive_response.get("Messages", [])

if messages:
    message = messages[0]
    ticket_data = json.loads(message["Body"])

    print("Received ticket:")
    print(json.dumps(ticket_data, indent=2))
else:
    print("No messages received.")
```

```python
if messages:
    sqs.delete_message(
        QueueUrl=queue_url,
        ReceiptHandle=message["ReceiptHandle"]
    )
    print("Ticket processed and message deleted.")
```

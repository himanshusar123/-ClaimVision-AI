```python
%%writefile quickcart_service.py
import csv

def check_return_eligibility(days, condition):
    if not isinstance(days, int):
        raise TypeError("Days must be an integer")
    if days < 0:
        raise ValueError("Days cannot be negative")
    if days > 30:
        return False
    return condition.strip().lower() == "unused"

def calculate_refund(amount, restocking_rate=0.05):
    if amount < 0:
        raise ValueError("Amount cannot be negative")
    if not 0 <= restocking_rate <= 1:
        raise ValueError("Rate must be between 0 and 1")
    return round(amount * (1 - restocking_rate), 2)

def create_return_record(order_id, customer_id, amount, status):
    if not order_id or not customer_id:
        raise ValueError("Order ID and customer ID are required")
    return {"order_id": str(order_id), "customer_id": str(customer_id),
            "refund_amount": round(float(amount), 2), "status": status}

def write_return_records(records, path):
    fields = ["order_id", "customer_id", "refund_amount", "status"]
    with open(path, "w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=fields)
        writer.writeheader()
        writer.writerows(records)
    return len(records)

def read_return_records(path):
    with open(path, "r", newline="", encoding="utf-8") as f:
        return list(csv.DictReader(f))
```

```python
%%writefile test_quickcart_service.py
import pytest
from quickcart_service import check_return_eligibility, calculate_refund, create_return_record

def test_return_eligible_within_30_days():
    assert check_return_eligibility(10, "unused") is True

def test_return_not_eligible_after_30_days():
    assert check_return_eligibility(35, "unused") is False

def test_used_item_is_not_eligible():
    assert check_return_eligibility(5, "used") is False

def test_refund_deducts_five_percent():
    assert calculate_refund(1000) == 950.00

def test_negative_amount_raises_error():
    with pytest.raises(ValueError, match="cannot be negative"):
        calculate_refund(-100)

def test_return_record_fields():
    row = create_return_record("ORD-501", "CUST-21", 950, "APPROVED")
    assert row["order_id"] == "ORD-501"
    assert row["customer_id"] == "CUST-21"
    assert row["refund_amount"] == 950.00
    assert row["status"] == "APPROVED"

def test_missing_order_id_raises_error():
    with pytest.raises(ValueError, match="required"):
        create_return_record("", "CUST-21", 950, "APPROVED")
```

```python
%%writefile test_quickcart_files.py
import pytest
from quickcart_service import write_return_records, read_return_records

@pytest.fixture
def sample_returns():
    return [
        {"order_id":"ORD-501","customer_id":"CUST-21","refund_amount":950.0,"status":"APPROVED"},
        {"order_id":"ORD-502","customer_id":"CUST-22","refund_amount":475.0,"status":"APPROVED"},
    ]

def test_write_and_read_csv(tmp_path, sample_returns):
    path = tmp_path / "returns.csv"
    count = write_return_records(sample_returns, path)
    rows = read_return_records(path)
    assert count == 2
    assert len(rows) == 2
    assert rows[0]["order_id"] == "ORD-501"
    assert rows[1]["status"] == "APPROVED"

def test_empty_csv(tmp_path):
    path = tmp_path / "empty.csv"
    assert write_return_records([], path) == 0
    assert read_return_records(path) == []
```

```python
%%writefile notification_service.py
def publish_return_notification(sqs_client, queue_url, record):
    response = sqs_client.send_message(
        QueueUrl=queue_url, MessageBody=str(record)
    )
    return {"success": True, "message_id": response["MessageId"]}

def safe_publish_return_notification(sqs_client, queue_url, record):
    try:
        return publish_return_notification(sqs_client, queue_url, record)
    except Exception as exc:
        return {"success": False, "error": str(exc)}
```

```python
%%writefile test_notification_service.py
from unittest.mock import Mock
from notification_service import safe_publish_return_notification

def test_notification_success():
    fake_sqs = Mock()
    fake_sqs.send_message.return_value = {"MessageId":"MSG-9001"}
    result = safe_publish_return_notification(
        fake_sqs, "https://example.test/quickcart-queue",
        {"order_id":"ORD-501","status":"APPROVED"})
    assert result["success"] is True
    assert result["message_id"] == "MSG-9001"
    fake_sqs.send_message.assert_called_once()

def test_notification_failure():
    fake_sqs = Mock()
    fake_sqs.send_message.side_effect = Exception("Queue unavailable")
    result = safe_publish_return_notification(
        fake_sqs, "https://example.test/quickcart-queue",
        {"order_id":"ORD-502","status":"APPROVED"})
    assert result["success"] is False
    assert result["error"] == "Queue unavailable"
    fake_sqs.send_message.assert_called_once()
```

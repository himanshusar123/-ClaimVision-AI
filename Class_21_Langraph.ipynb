```python
# Scenario: Order Processing Workflow in an E‑Commerce System
# Imagine an online store where every order goes through a sequence of steps:
# - Step 1 – Validation: The system checks if the order details are correct.
# - Step 2 – Payment: The system processes the payment.
# - Step 3 – Shipping: The system marks the order as ready for shipping.
# We can model this with a workflow graph so the order state flows through
#  each step automatically.


from langgraph.graph import StateGraph

# Step 1: Validate the order
# This function takes the current state (a dictionary),
# appends "-> validated" to the status, and returns the updated state.
def validate_order(state):
    state["status"] += " -> validated"
    return state

# Step 2: Process payment
# This function simulates payment processing by updating the status.
def process_payment(state):
    state["status"] += " -> payment processed"
    return state

# Step 3: Prepare for shipping
# This function marks the order as ready for shipping.
def prepare_shipping(state):
    state["status"] += " -> ready for shipping"
    return state

# Create the workflow graph
# StateGraph(dict) means our workflow will pass around a dictionary as state.
graph = StateGraph(dict)

# Add nodes (steps) to the graph
# Each node is a function that transforms the state.
graph.add_node("validate_order", validate_order)
graph.add_node("process_payment", process_payment)
graph.add_node("prepare_shipping", prepare_shipping)

# Connect nodes in sequence
# After validation, the workflow goes to payment,
# and after payment, it goes to shipping.
graph.add_edge("validate_order", "process_payment")
graph.add_edge("process_payment", "prepare_shipping")

# Set the entry point (starting node)
# The workflow begins at "validate_order".
graph.set_entry_point("validate_order")

# Compile the workflow into an executable pipeline
workflow = graph.compile()

# Run the workflow with an initial order state
# The state starts with "Order received".
result = workflow.invoke({"status": "Order received"})

# Print the final state after all steps are executed
print(result)
```

```python
# Scenario: Student Q&A Assistant
# Imagine a classroom assistant that helps students with basic questions.
# - A student asks: “What is AI?”
# - The assistant workflow processes the question and generates an answer.
# - Later, we could extend it with more nodes (like logging questions or
# checking difficulty), but for now we’ll keep it simple

# Import the StateGraph class from langgraph
# StateGraph lets us build workflows where "state" (data) flows through nodes (functions).
from langgraph.graph import StateGraph

# Step 1: Define the agent node function
# This function takes in a dictionary called "state".
# It looks for a key called "question", and then adds a new key called "answer".
def agent(state):
    # Extract the question from the state dictionary
    question = state["question"]

    # Provide a basic answer depending on the question
    # Here we hardcode a simple response for demonstration.
    if question.lower() == "what is ai?":
        state["answer"] = (
            "AI (Artificial Intelligence) is the field of computer science "
            "focused on building systems that can perform tasks requiring "
            "human-like intelligence, such as learning, reasoning, and problem-solving."
        )
    else:
        # If the question is not recognized, give a fallback response
        state["answer"] = f"Sorry, I don't have an answer for: {question}"

    # Return the updated state dictionary
    return state

# Step 2: Create the workflow graph
# We tell StateGraph that our state will be represented as a Python
#  dictionary.
graph = StateGraph(dict)

# Step 3: Add the agent node to the graph
# "agent" is the name of the node, and agent (the function) is what
#  runs at that node.
graph.add_node("agent", agent)

# Step 4: Set the entry point
# This tells the workflow where to start. In this case, it starts at the "agent" node.
graph.set_entry_point("agent")

# Step 5: Compile the workflow
# This turns the graph definition into an executable pipeline.
workflow = graph.compile()

# Step 6: Run the workflow
# We invoke the workflow with an initial state dictionary containing a
# question.
result = workflow.invoke({"question": "What is AI?"})

# Step 7: Print the result
# The result will include both the original question and the generated answer.
print(result)
```

```python
# AI Code Assistant Agent
# Scenario

# A coding assistant:

# reads user request

# generates code

# explains code

from langgraph.graph import StateGraph

def understand_request(state):
    request = state["request"]
    state["language"] = "Python"
    return state

def generate_code(state):
    state["code"] = "def add(a,b): return a+b"
    return state

def explain_code(state):
    state["explanation"] = "This function adds two numbers."
    return state

graph = StateGraph(dict)

graph.add_node("understand", understand_request)
graph.add_node("generate", generate_code)
graph.add_node("explain", explain_code)

graph.set_entry_point("understand")

graph.add_edge("understand", "generate")
graph.add_edge("generate", "explain")

workflow = graph.compile()

result = workflow.invoke({"request": "Write a function to add numbers"})

print(result)
```
